```python
# CELL 6: Inspect Generated Artifacts
import json
from pathlib import Path

onnx_path = Path('models/annadata_production.onnx')
cls_path  = Path('artifacts/disease_classes.json')

print(f'ONNX model : {onnx_path}  ({onnx_path.stat().st_size/1e6:.2f} MB)')

with open(cls_path) as f:
    cls = json.load(f)
print(f'Classes    : {len(cls)}')
print('Sample class mappings:')
for k in ['0','15','29','37']:
    print(f'  [{k:>2}] {cls[k]["name"]}')
```
