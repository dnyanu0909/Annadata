# Annadata API Server on Colab

Launches the FastAPI backend on the Colab runtime and exposes it
publicly via an ngrok tunnel so your mobile app or web frontend can reach it.

**Prerequisites:** Run `Annadata_Colab.ipynb` first to train the model and download artifacts.

```python
# CELL 2: Clone repo (if not already present) and confirm model artifacts exist
import os
from pathlib import Path

if not Path('/content/Annadata').exists():
    !git clone https://github.com/dnyanu0909/Annadata.git

%cd /content/Annadata

# Verify model artifacts exist (copy from your Colab session or upload manually)
onnx = Path('models/annadata_production.onnx')
cls  = Path('artifacts/disease_classes.json')

assert onnx.exists(), f'Missing: {onnx}  -- Upload it via the Files panel or rerun training.'
assert cls.exists(),  f'Missing: {cls}   -- Upload it via the Files panel or rerun training.'

print(f'ONNX model  : {onnx.stat().st_size/1e6:.2f} MB')
print(f'Class map   : {cls.stat().st_size/1024:.1f} KB')
print('Artifacts verified.')
```

```python
# CELL 3: Configure ngrok (free public tunnel)
# Get your free authtoken at: https://dashboard.ngrok.com/get-started/your-authtoken
NGROK_AUTHTOKEN = 'your_ngrok_authtoken_here'  # <-- paste your ngrok token

from pyngrok import ngrok, conf
conf.get_default().auth_token = NGROK_AUTHTOKEN
print('ngrok configured.')
```

```python
# CELL 4: Start the FastAPI server and expose it publicly
import subprocess, threading, time, requests
from pyngrok import ngrok

# Kill any previous uvicorn processes
!pkill -f uvicorn 2>/dev/null || true
time.sleep(1)

# Start uvicorn in background
server = subprocess.Popen(
    ['python', '-m', 'uvicorn', 'api.main:app', '--host', '0.0.0.0', '--port', '8000'],
    cwd='/content/Annadata',
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT
)

# Wait for server to be ready
for _ in range(20):
    try:
        r = requests.get('http://localhost:8000/', timeout=1)
        if r.status_code == 200:
            break
    except Exception:
        time.sleep(1)

# Open ngrok tunnel
tunnel = ngrok.connect(8000)
public_url = tunnel.public_url

print()
print('=' * 60)
print('  Annadata API is LIVE')
print('=' * 60)
print(f'  Public URL : {public_url}')
print(f'  Swagger UI : {public_url}/docs')
print(f'  Health     : {public_url}/')
print(f'  Predict    : POST {public_url}/predict')
print('=' * 60)
```

```python
# CELL 5: Quick smoke test against the live API
import requests, json
from pathlib import Path

# Health check
health = requests.get(f'{public_url}/').json()
print('Health:', json.dumps(health, indent=2))

# Predict with a real image
test_img = Path('/content/Annadata/test_leaf.jpg')
if test_img.exists():
    with open(test_img, 'rb') as f:
        response = requests.post(
            f'{public_url}/predict?top_k=5',
            files={'file': ('test_leaf.jpg', f, 'image/jpeg')}
        )
    print()
    print('Prediction:', json.dumps(response.json(), indent=2))
else:
    print('No test image found -- upload one and re-run this cell.')
```
